```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np
```

```python
data = {
    'Study_Hours': [1, 2, 3, 4, 5],
    'Marks': [40, 50, 55, 70, 75]
}

df = pd.DataFrame(data)
```

```python
X = df[['Study_Hours']]
y = df['Marks']
```

```python
# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict
y_pred = model.predict(X)

# Predict marks for 6 study hours
prediction = model.predict(pd.DataFrame([[6]], columns=['Study_Hours']))
```

```python
print("Predicted marks for 6 hours:", prediction[0])
```

```python
# R2 score
r2 = r2_score(y, y_pred)
print("R2 Score:", r2)

# RMSE
rmse = np.sqrt(mean_squared_error(y, y_pred))
print("RMSE:", rmse)
```

```python
# Regression line
plt.scatter(X, y)
plt.plot(X, y_pred)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Simple Linear Regression")

plt.show()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np
```

```python
data = {
    'Study_Hours': [2, 3, 4, 5, 6],
    'Attendance': [70, 75, 80, 85, 90],
    'Previous_Marks': [50, 55, 60, 65, 70],
    'Final_Marks': [55, 60, 68, 75, 82]
}

df = pd.DataFrame(data)
```

```python
# Input variables
X = df[['Study_Hours', 'Attendance', 'Previous_Marks']]

# Output variable
y = df['Final_Marks']
```

```python
# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict existing data
y_pred = model.predict(X)

# Predict for a new student
new_student = pd.DataFrame(
    [[6, 90, 70]],
    columns=['Study_Hours', 'Attendance', 'Previous_Marks']
)
```

```python
prediction = model.predict(new_student)
print("Predicted Final Marks:", prediction[0])
rmse = np.sqrt(mean_squared_error(y, y_pred))
print("RMSE:", rmse)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Input data: Study Hours
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)

# Output data: Marks
y = np.array([40, 50, 55, 70, 75])
```

```python
# Create the model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Display slope and intercept
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)

# Predict marks for 6 study hours
prediction = model.predict([[6]])
```

```python
print("Predicted Marks:", prediction[0])
```

```python
# Draw actual data points
plt.scatter(X, y, color="blue", label="Actual Marks")

# Draw regression line
plt.plot(X, model.predict(X),color="red", label="Regression Line")
plt.scatter(6, prediction[0],
            color="green",
            s=100,
            marker="*",
            label="Predicted Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Student Marks")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Input data: Study Hours
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)

# Actual output: Marks
y = np.array([40, 50, 55, 70, 75])
```

```python
# Create the model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict marks for the original students
y_pred = model.predict(X)
```

```python
# Calculate R-squared
r2 = r2_score(y, y_pred)

# Calculate RMSE
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
```

```python
# Display evaluation results
print("R-squared:", r2)
print("R-squared Percentage:", r2 * 100, "%")
print("RMSE:", rmse)

# Predict marks for a new student
prediction = model.predict([[6]])
print("Predicted Marks for 6 Hours:", prediction[0])
```

```python
# Plot actual data and regression line
plt.scatter(X, y, color="blue", label="Actual Marks")
plt.plot(X, y_pred, color="red", label="Regression Line")

# Show the new predicted point
plt.scatter(6, prediction[0], color="green",
            marker="*", s=150, label="New Prediction")

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Simple Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```
