```python
import pandas as pd
import numpy as np
```

```python
data={"Student":["A","B","C","D","E"],
      "Gender":["Male","Female","Male","Female","Male"],
      "StudyHours":[5,6,np.nan,4,7],
      "Attendance":[80,85,75,np.nan,90],
      "Marks":[75,72,68,55,np.nan],
      "Result":["Pass","Pass","Pass","Fail","Pass"]}
```

```python
df=pd.DataFrame(data)
```

```python
print(df)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df["StudyHours"]=df["StudyHours"].fillna(df["StudyHours"].mean())
print("\nData set after mean imputation:")
print(df)
```

```python
df["Marks"]=df["Marks"].fillna(df["Marks"].median())
print("\nData set after median imputation:")
print(df)
```

```python
df["Attendance"]=df["Attendance"].ffill()
print("\nData set after Forward fill:")
print(df)
```

```python
print(df["Gender"].unique())
df["Gender"]=df["Gender"].map({"Male":1,"Female":0})
print("\nData set after gender enoding:")
print(df)
```

```python
df["Result"]=df["Result"].map({"Pass":1,"Fail":0})
print("\nData set after result encoding:")
print(df)
```

```python
print("\nFinal preprocessed data set:")
print(df)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```
