```python
import pandas as pd
import numpy as np
```

```python
data={"Students":["A","B","C","D","E","F","G","H","I","J"],"Marks":[50,55,60,62,65,68,70,72,75,150]}
df=pd.DataFrame(data)
```

```python
Q1=df["Marks"].quantile(0.25)
print(Q1)
```

```python
Q3=df["Marks"].quantile(0.75)
print(Q3)
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lowerlimit=Q1-(1.5*IQR)
upperlimit=Q3+(1.5*IQR)
print("\nQ1:",Q1)
print("\nQ3:",Q3)
print("\nIQR:",IQR)
print("\nLower Limit:",lowerlimit)
print("\nUpper Limit:",upperlimit)
```

```python
outliers=df[(df["Marks"]<lowerlimit)|(df["Marks"]>upperlimit)]
```

```python
print("\nOutliers using IQR:")
print(outliers)
```

```python
df_capped=df.copy()
df_capped["Marks"]=df_capped["Marks"].clip(lower=lowerlimit,upper=upperlimit)
print("\nAfter Capping:")
print(df_capped)
```

```python
df["Marks_Treated"]=np.where(df["Marks"]>upperlimit,upperlimit,np.where(df["Marks"]<lowerlimit,lowerlimit,df["Marks"]))
```
