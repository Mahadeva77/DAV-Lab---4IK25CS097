```python
from google.colab import files
uploaded = files.upload()
import pandas as pd
data=pd.read_csv('agri (1).csv')
df=pd.DataFrame(data)
print(df)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df["Area_Hectares"]=df["Area_Hectares"].fillna(df["Area_Hectares"].mean())
print("\nData set after mean imputation:")
print(df)
```

```python
df["Temperature_C"]=df["Temperature_C"].fillna(df["Temperature_C"].median())
print("\nData set after median imputation:")
print(df)
```

```python
df["Rainfall_mm"]=df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())
print("\nData set after median imputation:")
print(df)
```

```python
df["Production_Tonnes"]=df["Production_Tonnes"].ffill()
print("\nData set after Forward fill:")
print(df)
```

```python
print(df["Season"].unique())
df["Season"]=df["Season"].map({"Kharif":1,"Rabi":0,"Annual":2})
print("\nData set after season enoding:")
print(df)
```

```python
print("\nPre-processed data set:")
print(df)
```
